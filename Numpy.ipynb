<b>Time for action – stacking arrays

```python
a = np.arange(9).reshape(3,3)
a
```

```python
b = 2 * a
b
```

```python
np.hstack((a, b))
```

```python
np.concatenate((a, b), axis=1)
```

```python
np.vstack((a, b))
```

```python
np.concatenate((a, b), axis=0)
```

```python
np.dstack((a, b))   #depth stacking
```

```python
oned = np.arange(2)
oned
```

```python
twice_oned = 2 * oned
twice_oned
```

```python
np.column_stack((oned, twice_oned))
```

```python
np.column_stack((a, b))   #similar to hstack
```

```python
np.column_stack((a, b)) == np.hstack((a, b))
```

```python
np.row_stack((oned, twice_oned))  #similar to vstack
```

```python
np.row_stack((a, b))
```

```python
np.row_stack((a,b)) == np.vstack((a, b))
```
